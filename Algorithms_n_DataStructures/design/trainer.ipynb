```python
class ClassMeth:
    name = "cls_name"
    
    def __init__(self, val):
        self.val = val
        self._protected = "protected"
        self.__private = "private"
    
    @classmethod
    def print_method(cls):
        print(f'{cls=}, {cls.name=}')
    
    @staticmethod
    def calc(n):
        return n ** 2
```

```python
ClassMeth
```

```python
obj = ClassMeth(10)
```

```python
obj
```

```python
ClassMeth.print_method()
```
