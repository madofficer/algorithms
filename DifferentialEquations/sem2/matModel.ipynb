```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
e1 = 0.7
e2 = 0.1
y1 = 0.07
y2 = 0.07
epsilon = 0.01
```

```python
x0 = 40
y0 = 9
```

```python
phi = np.arctan((y0 - (e1 / y2)) / (x0 - (e2 / y1)))
a = (x0 - (e2 / y1))/ np.cos(phi)
phi, a
```

```python
t = np.linspace(0, 100, 1000)
```

```python
omega1 = e2 * y2 / y1
omega2 = y1 * e1 / y2

N1 = (
    (e2 / y1) +
    a * np.cos(omega1 * t + phi) +
    epsilon * (
        -a**2 / 2 * (np.sin(2 * omega1 * t + 2 * phi) - np.sin(2 * phi)) +
        a**2 / 4 * (np.cos(2 * omega1 * t + 2 * phi) - np.cos(2 * phi))
    )
)

N2 = (
    (e1 / y2) +
    a * np.sin(omega2 * t + phi) +
    epsilon * (
        a**2 / 2 * (np.sin(2 * omega2 * t + 2 * phi) - np.sin(2 * phi)) -
        a**2 / 4 * (np.cos(2 * omega2 * t + 2 * phi) - np.cos(2 * phi))
    )
)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(t, N1, label='$N_1(t)$', color='blue')
plt.plot(t, N2, label='$N_2(t)$', color='orange')
plt.title('Графики $N_1(t)$ и $N_2(t)$')
plt.xlabel('Время $t$')
plt.ylabel('Численность')
plt.legend()
plt.grid(True)
plt.show()
```
