```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Параметры функции
A = 0.095       # Амплитуда
omega = 6.63    # Угловая частота

# Создаем массив времени t от 0 до 2 секунд с шагом 0.01
t = np.arange(0, 10, 0.01)

# Вычисляем значения x(t)
x = A * np.sin(omega * t)
```

```python


# Строим график
plt.figure(figsize=(10, 5))
plt.plot(t, x, label=r'$x(t) = 0.095 \cdot \sin(6.63t)$', color='blue')
plt.title('График колебаний пружины')
plt.xlabel('Время t, с')
plt.ylabel('Смещение x(t), м')
plt.grid(True)
plt.legend()
plt.show()
```
