```python
import numpy as np
```

```python
# graphical
x1, y1 = 2.20293, 0.092872
x2, y2 = 2.50406, 0.165059
```

```python
x2 - x1
```

```python
tg = (y2 - y1) / (x2 - x1)
tg
```

```python
rad = np.arctan(tg)
rad
```

```python
deg = np.degrees(rad)
deg
```

```python
# analytical
gamma, p0, ro0 = 1.41, 101325, 1.17666
v_s = np.sqrt(gamma * p0 / ro0)
v0 = 1041.27
M = v0 / v_s
```

```python
M
```

```python
M_rad = np.arcsin(1 / M)
M_rad
```

```python
M_deg = np.degrees(M_rad)
M_deg
```
