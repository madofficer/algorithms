```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x) -> float:
    return x ** 3 - 5 * x ** 2 + 4.5
```

```python
def f_prime(x):
    return 3 * x ** 2 - 10 * x
```

```python
def newton_method(f, f_prime, x0, epsilon):
    x_n = x0
    iters = [x_n]
    errors = []
    
    while True:
        x_next = x_n - f(x_n) / f_prime(x_n)
        iters.append(x_next)
        error = abs(x_next - x_n)
        errors.append(error)
        
        if error < epsilon:
            break
    
        x_n = x_next
        
    return iters, errors
```

```python
x0 = 4.7 # начальное приближение
epsilon_1 = 1e-3
epsilon_2 = 1e-6
```

```python
iterations_1, errors_1 = newton_method(f, f_prime, x0, epsilon_1)
iterations_2, errors_2 = newton_method(f, f_prime, x0, epsilon_2)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(iterations_1, 'bo-', label='Приближения')
plt.title('Метод Ньютона (\u03B5 = 1e-3)')
plt.xlabel('Итерации')
plt.ylabel('x')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(iterations_2, 'ro-', label='Приближения')
plt.title('Метод Ньютона (\u03B5 = 1e-6)')
plt.xlabel('Итерации')
plt.ylabel('x')
plt.grid(True)

plt.tight_layout()
plt.show()
```
